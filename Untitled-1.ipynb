```python
import numpy as np
import matplotlib.pyplot as plt

dx = 0.2
x = np.arange(0, 30 + dx, dx)
y = np.zeros(len(x))

##-----------------------------
##Forward Euler Implementation
##-----------------------------
alpha = 1
y[0] = 1
for i in range(len(x) - 1):
    y[i + 1] = y[i] + dx * (-alpha * y[i])
##------------------------------

y_exact = np.exp(-alpha * x)
plt.plot(x, y_exact)
plt.plot(x, y)
plt.legend(["exact solution", "Forward Euler solution"])
```
